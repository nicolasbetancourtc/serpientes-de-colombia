```python
import os
```

```python
folder = r"C:\\Users\\nicolas.betancourt\\Documents\\GitHub\\pytorch\\serpientes-de-colombia\\data\\01_raw\\serpientes_de_colombia_images\\falsa_vibora"  

for filename in os.listdir(folder):
    if filename.lower().endswith(".jpg.jpg"):
        old_path = os.path.join(folder, filename)
        new_filename = filename[:-4]  # remove the last ".jpg"
        new_path = os.path.join(folder, new_filename)
        os.rename(old_path, new_path)
        print(f"Renamed: {filename} → {new_filename}")
```
